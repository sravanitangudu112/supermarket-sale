## Data Cleaning

```python
# Strip whitespace from string columns and replace non-breaking spaces
for col in df.select_dtypes(include=['object']).columns:
    df[col] = df[col].str.strip()
    df[col] = df[col].str.replace('\u00a0', ' ', regex=False)

print("Cleaned string columns:")
print(df.head())
```

```python
# Check for duplicate rows
duplicates_count = df.duplicated().sum()
if duplicates_count > 0:
    print(f"Found {duplicates_count} duplicate rows. Removing them...")
    df.drop_duplicates(inplace=True)
    print("Duplicates removed. New shape:", df.shape)
else:
    print("No duplicate rows found.")
```

Let's perform some mathematical operations on a numerical column using NumPy. We will use the 'Item Code' column for this demonstration.

```python
# Extract the 'Item Code' column as a NumPy array
item_code_array = df['Item Code'].to_numpy()

# Perform mathematical operations
mean_item_code = np.mean(item_code_array)
min_item_code = np.min(item_code_array)
max_item_code = np.max(item_code_array)
std_item_code = np.std(item_code_array)

print(f"Mean of Item Code: {mean_item_code}")
print(f"Minimum Item Code: {min_item_code}")
print(f"Maximum Item Code: {max_item_code}")
print(f"Standard Deviation of Item Code: {std_item_code}")
```

Let's explore indexing and slicing of the `item_code_array` using NumPy.

```python
# Accessing a single element (e.g., the first element)
first_element = item_code_array[0]
print(f"First element of Item Code array: {first_element}")

# Accessing a specific element (e.g., the 5th element)
specific_element = item_code_array[4] # Index 4 for the 5th element
print(f"Fifth element of Item Code array: {specific_element}")

# Slicing the array (e.g., first 5 elements)
first_five_elements = item_code_array[0:5]
print(f"\nFirst five elements: {first_five_elements}")

# Slicing from a specific index to the end (e.g., from index 10 onwards)
from_index_ten = item_code_array[10:]
print(f"\nElements from index 10 onwards (first 5 shown): {from_index_ten[:5]}...")

# Slicing with a step (e.g., every other element for the first 10)
every_other_element = item_code_array[0:10:2]
print(f"\nEvery other element (first 10, every other): {every_other_element}")

# Reverse the array
reversed_array = item_code_array[::-1]
print(f"\nReversed array (first 5 shown): {reversed_array[:5]}...")
```

Let's perform a univariate analysis for the 'Category Name' column using a countplot to visualize the distribution of different categories.

```python
plt.figure(figsize=(12, 7))
sns.countplot(data=df, y='Category Name', order=df['Category Name'].value_counts().index, palette='viridis')
plt.title('Count of Items per Category')
plt.xlabel('Count')
plt.ylabel('Category Name')
plt.grid(axis='x', alpha=0.75)
plt.tight_layout()
plt.show()
```

Let's perform a bivariate analysis for 'Item Code' and 'Item Name'. Since there are many unique item names, we'll focus on the top 10 most frequent ones for better readability in the plots.

```python
# Get the top 10 most frequent Item Names
top_10_item_names = df['Item Name'].value_counts().nlargest(10).index

# Filter the DataFrame for these top items
df_top_items = df[df['Item Name'].isin(top_10_item_names)]

# Create a bar chart for the count of top 10 Item Names
plt.figure(figsize=(12, 7))
sns.countplot(data=df_top_items, y='Item Name', order=top_10_item_names, palette='coolwarm')
plt.title('Top 10 Most Frequent Item Names')
plt.xlabel('Count')
plt.ylabel('Item Name')
plt.grid(axis='x', alpha=0.75)
plt.tight_layout()
plt.show()
```

```python
# Create a box plot for Item Code vs. Item Name for the top 10 Item Names
# Note: If Item Code is unique for each Item Name, the 'box' will be just a single line.
plt.figure(figsize=(12, 7))
sns.boxplot(data=df_top_items, y='Item Name', x='Item Code', order=top_10_item_names, palette='viridis')
plt.title('Item Code Distribution for Top 10 Item Names')
plt.xlabel('Item Code')
plt.ylabel('Item Name')
plt.grid(axis='x', alpha=0.75)
plt.tight_layout()
plt.show()
```

Let's identify the 'best-selling' product by finding the item name that appears most frequently in the dataset.

```python
# Count the occurrences of each 'Item Name'
item_name_counts = df['Item Name'].value_counts()

# Get the most frequent item
best_selling_product = item_name_counts.idxmax()
best_selling_count = item_name_counts.max()

print(f"The best-selling product (most frequent) is: '{best_selling_product}'")
print(f"It appears {best_selling_count} times in the dataset.")
```

### Insight:
Based on the frequency of items in the dataset, **'Chinese Cabbage'** and **'Broccoli'** are tied as the most frequently listed products, each appearing 2 times. This suggests they might be popular items in terms of their variety or stock keeping within the supermarket's inventory data provided.

Now, let's look at the 'least bought' items, interpreting this as the items that appear least frequently in our dataset.

```python
# Find the minimum count
least_frequent_count = item_name_counts.min()

# Get all item names that have this minimum count
least_bought_products = item_name_counts[item_name_counts == least_frequent_count].index.tolist()

print(f"The least bought products (least frequent, appearing {least_frequent_count} time(s)) are:")
for product in least_bought_products:
    print(f"- '{product}'")
```

### Insights on Least Bought Items:

Similar to the 'best-selling' analysis, our understanding of 'least bought' is limited by the available data. Since many items appear only once, it's difficult to draw strong conclusions.

*   **High Variety, Low Frequency**: The large number of items appearing only once suggests a wide variety of products in the inventory, many of which may not have high individual transaction volumes. In a real-world scenario, these could be niche products, seasonal items, or new arrivals that haven't gained traction.
*   **Potential for Removal or Promotion**: For a supermarket, items with consistently low sales (which could be reflected by low frequency in a comprehensive sales dataset) might be candidates for removal from shelves to optimize space, or for targeted promotions to boost their sales.
*   **Data Collection Implications**: If this dataset truly represents sales, then the single occurrences could indicate items that are not selling well. However, without actual sales quantities or revenue, we cannot determine if these items are truly 'unpopular' or simply have a lower turnover due to their nature (e.g., expensive specialty items bought less frequently).

**To gain actionable insights for these least bought items, a more robust sales dataset with actual transaction data, sales volumes, dates, and potentially pricing information would be crucial.** This would allow for analysis such as:
*   Identifying dead stock.
*   Understanding why certain products don't sell (e.g., poor placement, high price, lack of demand).
*   Developing strategies for clearance, bundle offers, or discontinuation.
